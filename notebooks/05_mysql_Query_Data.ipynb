```python
import datetime
import mysql.connector
```

```python
cnx = mysql.connector.connect(user='root', database='accidents')
cursor = cnx.cursor()

query = (
  "SELECT operator, COUNT(*) AS operator_count "
  "FROM accidents "
  "GROUP BY operator "
  "ORDER BY operator_count DESC "
  "LIMIT 10"
)

cursor.execute(query)

for operator, operator_count in cursor:
  print(f"{operator}: {operator_count}")

cursor.close()
cnx.close()
```

```python
cnx = mysql.connector.connect(user='root', database='accidents')
cursor = cnx.cursor()

query = (
  "SELECT ac_type, COUNT(*) AS ac_count "
  "FROM accidents "
  "GROUP BY ac_type "
  "ORDER BY ac_count DESC "
  "LIMIT 10"
)

cursor.execute(query)

print("Top 10 accident counts by aircraft type")
for ac_type, ac_count in cursor:
  print(f"{ac_type}: {ac_count}")

cursor.close()
cnx.close()
```

```python
cnx = mysql.connector.connect(user='root', database='accidents')
cursor = cnx.cursor()

# Create a query to get number of accidents happened by Douglas Dc-3 each year
query = (
  "SELECT year, COUNT(*) AS ac_count "
  "FROM accidents "
  "WHERE ac_type = 'Douglas DC-3' "
  "GROUP BY year "
  "ORDER BY ac_count DESC"
)
cursor.execute(query)

print("Number of accidents by year for Douglas DC-3")
for year, ac_count in cursor:
  print(f"{year}: {ac_count}")

cursor.close()
cnx.close()
```
